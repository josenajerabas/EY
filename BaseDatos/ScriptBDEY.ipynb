# [dbo].[sp_ActualizarUsuario]

```sql
/****** Object:  StoredProcedure [dbo].[sp_ActualizarUsuario]    Script Date: 15/02/2026 11:52:52 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[sp_ActualizarUsuario]
    @Id INT,
    @NombreCompleto NVARCHAR(200),
    @NombreUsuario NVARCHAR(50),
    @Password NVARCHAR(100) = NULL, -- Opcional al editar
    @Correo NVARCHAR(100),
    @Estatus BIT
AS
BEGIN
    SET NOCOUNT ON;
    
    BEGIN TRY
        BEGIN TRANSACTION;
        
        -- Verificar que el usuario existe
        IF NOT EXISTS (SELECT 1 FROM Usuarios WHERE Id = @Id)
        BEGIN
            RAISERROR('Usuario no encontrado', 16, 1);
            RETURN;
        END
        
        -- Verificar nombre de usuario duplicado (excluyendo el actual)
        IF EXISTS (SELECT 1 FROM Usuarios WHERE NombreUsuario = @NombreUsuario AND Id <> @Id)
        BEGIN
            RAISERROR('El nombre de usuario ya existe', 16, 1);
            RETURN;
        END
        
        -- Verificar correo duplicado (excluyendo el actual)
        IF EXISTS (SELECT 1 FROM Usuarios WHERE Correo = @Correo AND Id <> @Id)
        BEGIN
            RAISERROR('El correo electrónico ya está registrado', 16, 1);
            RETURN;
        END
        
        -- Actualizar usuario
        IF @Password IS NOT NULL AND LEN(@Password) > 0
        BEGIN
            -- Actualizar con nueva contraseña
            UPDATE Usuarios
            SET NombreCompleto = @NombreCompleto,
                NombreUsuario = @NombreUsuario,
                Password = @Password,
                Correo = @Correo,
                Estatus = @Estatus,
                FechaModificacion = GETDATE()
            WHERE Id = @Id;
        END
        ELSE
        BEGIN
            -- Actualizar sin cambiar contraseña
            UPDATE Usuarios
            SET NombreCompleto = @NombreCompleto,
                NombreUsuario = @NombreUsuario,
                Correo = @Correo,
                Estatus = @Estatus,
                FechaModificacion = GETDATE()
            WHERE Id = @Id;
        END
        
        COMMIT TRANSACTION;
    END TRY
    BEGIN CATCH
        IF @@TRANCOUNT > 0
            ROLLBACK TRANSACTION;
        
        THROW;
    END CATCH
END
GO
```

# [dbo].[sp_EliminarUsuario]

```sql
/****** Object:  StoredProcedure [dbo].[sp_EliminarUsuario]    Script Date: 15/02/2026 11:52:52 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[sp_EliminarUsuario]
    @Id INT
AS
BEGIN
    SET NOCOUNT ON;
    
    BEGIN TRY
        BEGIN TRANSACTION;
        
        -- Verificar que el usuario existe
        IF NOT EXISTS (SELECT 1 FROM Usuarios WHERE Id = @Id)
        BEGIN
            RAISERROR('Usuario no encontrado', 16, 1);
            RETURN;
        END
        
        -- Eliminar usuario
        DELETE FROM Usuarios WHERE Id = @Id;
        
        COMMIT TRANSACTION;
    END TRY
    BEGIN CATCH
        IF @@TRANCOUNT > 0
            ROLLBACK TRANSACTION;
        
        THROW;
    END CATCH
END
GO
```

# [dbo].[sp_InsertarUsuario]

```sql
/****** Object:  StoredProcedure [dbo].[sp_InsertarUsuario]    Script Date: 15/02/2026 11:52:52 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[sp_InsertarUsuario]
    @NombreCompleto NVARCHAR(200),
    @NombreUsuario NVARCHAR(50),
    @Password NVARCHAR(100),
    @Correo NVARCHAR(100),
    @Estatus BIT = 1,
    @UsuarioId INT OUTPUT
AS
BEGIN
    SET NOCOUNT ON;
    
    BEGIN TRY
        BEGIN TRANSACTION;
        
        -- Verificar si el usuario ya existe
        IF EXISTS (SELECT 1 FROM Usuarios WHERE NombreUsuario = @NombreUsuario)
        BEGIN
            RAISERROR('El nombre de usuario ya existe', 16, 1);
            RETURN;
        END
        
        -- Verificar si el correo ya existe
        IF EXISTS (SELECT 1 FROM Usuarios WHERE Correo = @Correo)
        BEGIN
            RAISERROR('El correo electrónico ya está registrado', 16, 1);
            RETURN;
        END
        
        -- Insertar usuario
        INSERT INTO Usuarios (NombreCompleto, NombreUsuario, Password, Correo, Estatus, FechaAlta)
        VALUES (@NombreCompleto, @NombreUsuario, @Password, @Correo, @Estatus, GETDATE());
        
        -- Obtener ID del usuario insertado
        SET @UsuarioId = SCOPE_IDENTITY();
        
        COMMIT TRANSACTION;
    END TRY
    BEGIN CATCH
        IF @@TRANCOUNT > 0
            ROLLBACK TRANSACTION;
        
        THROW;
    END CATCH
END
GO
```

# [dbo].[sp_ObtenerEstadisticasDashboard]

```sql
/****** Object:  StoredProcedure [dbo].[sp_ObtenerEstadisticasDashboard]    Script Date: 15/02/2026 11:52:52 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[sp_ObtenerEstadisticasDashboard]
AS
BEGIN
    SET NOCOUNT ON;
    
    SELECT 
        -- Total de usuarios
        COUNT(*) AS TotalUsuarios,
        
        -- Usuarios activos
        SUM(CASE WHEN Estatus = 1 THEN 1 ELSE 0 END) AS UsuariosActivos,
        
        -- Usuarios inactivos
        SUM(CASE WHEN Estatus = 0 THEN 1 ELSE 0 END) AS UsuariosInactivos,
        
        -- Registrados hoy
        SUM(CASE WHEN CAST(FechaAlta AS DATE) = CAST(GETDATE() AS DATE) THEN 1 ELSE 0 END) AS UsuariosRegistradosHoy,
        
        -- Registrados esta semana
        SUM(CASE WHEN FechaAlta >= DATEADD(DAY, -7, GETDATE()) THEN 1 ELSE 0 END) AS UsuariosRegistradosSemana,
        
        -- Registrados este mes
        SUM(CASE WHEN FechaAlta >= DATEADD(DAY, -30, GETDATE()) THEN 1 ELSE 0 END) AS UsuariosRegistradosMes
    FROM Usuarios;
END
GO
```

# [dbo].[sp_ObtenerRegistrosPorDia]

```sql
/****** Object:  StoredProcedure [dbo].[sp_ObtenerRegistrosPorDia]    Script Date: 15/02/2026 11:52:52 p. m. ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE PROCEDURE [dbo].[sp_ObtenerRegistrosPorDia]
AS
BEGIN
    SET NOCOUNT ON;
    
    SELECT 
        CONVERT(VARCHAR(10), FechaAlta, 103) AS Fecha,
        COUNT(*) AS Cantidad
    FROM Usuarios
    WHERE FechaAlta >= DATEADD(DAY, -30, GETDATE())
    GROUP BY CAST(FechaAlta AS DATE), CONVERT(VARCHAR(10), FechaAlta, 103)
    ORDER BY CAST(FechaAlta AS DATE) ASC;
END
GO
```
